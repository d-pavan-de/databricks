```python
input_path = "/Volumes/dbacademy/bronze/autoloader/input2/"
checkpoint_path = "/Volumes/dbacademy/bronze/autoloader/process/"
```

```python

df = (spark.readStream
.format("cloudFiles")
.option("cloudFiles.format", "parquet")
.option("cloudFiles.schemaLocation", checkpoint_path)
.option("cloudFiles.inferSchema", "true")
#.option("cloudFiles.schemaEvolutionMode", "rescue")
.load(input_path)
)
```

```python
(df.writeStream
    .trigger(availableNow=True)
    .option("checkpointLocation", checkpoint_path)
    .toTable("dbacademy.bronze.sql_parquet_autoloader")
)
```
