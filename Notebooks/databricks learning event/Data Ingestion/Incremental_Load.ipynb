### COPY INTO : load data incrementally
You can use the COPY INTO SQL command to load data from a file location into a Delta table. COPY INTO is retriable and idempotent — files in the source location that have already been loaded are skipped on subsequent runs.

COPY INTO offers these capabilities:

Easily configurable file or folder filters from cloud storage, including S3, ADLS, ABFS, GCS, and Unity Catalog volumes.
Support for multiple source file formats: CSV, JSON, XML, Avro, ORC, Parquet, text, and binary files.
Exactly-once (idempotent) file processing by default.
Target table schema inference, mapping, merging, and evolution.

```sql
use dbacademy.bronze;
```

```sql
-- drop table if exists
drop table if exists cp_into_orders;

-- create an empty table
CREATE TABLE cp_into_orders;

-- LOAD data using copy into
copy into cp_into_orders
from '/Volumes/databrickscert/databrickscert/samplevolume/parquet/orders.parquet'
FILEFORMAT = PARQUET
copy_options  ('mergeSchema'='true');
```

```sql

describe table extended cp_into_orders;
```

```sql
-- view data from cp_into_orders
select * from dbacademy.bronze.cp_into_orders limit 10
```

```sql
-- drop table if exists
drop table if exists cp_from_csv_sales;

--create an empty table
CREATE TABLE cp_from_csv_sales;

describe table extended cp_from_csv_sales;
```

```sql
-- IDEMPOTENCY running this cell 2nd time will add copy any data idempotency 
copy into cp_from_csv_sales
from (select product, quantity, price from '/Volumes/databrickscert/databrickscert/samplevolume/CSV/sales.csv')
FILEFORMAT = CSV
--PATTERN = 'folder1/file_[a-g].csv' -- pattern can be used to load only specific files
FORMAT_OPTIONS('header' = 'true')
COPY_OPTIONS ('mergeSchema'='true');
```

```sql
select * from cp_from_csv_sales limit 10
```
