```sql
drop  table if exists orders;

-- create the delta table
create table orders as
select *, _metadata.file_name as filename, _metadata.file_modification_time as file_modification_time from read_files(
    '/Volumes/databrickscert/databrickscert/samplevolume/parquet/orders.parquet',
    format=>'parquet'
);

-- Read orders table
select * from orders limit 10;
```

```sql
%python
# Read file into a Spark DataFrame
df = spark.read.format("parquet").load("/Volumes/databrickscert/databrickscert/samplevolume/parquet/orders.parquet")

# write a delta table
df.write.mode("overwrite").saveAsTable("pyspark_orders")

#display the table data
df1= spark.table("pyspark_orders").limit(10)
display(df1)
```
