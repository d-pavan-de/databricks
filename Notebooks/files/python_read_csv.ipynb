### read a CSV using heder option & inferschema

```sql
%python

df = spark.read.format("csv")\
    .option("header", "True")\
    .option("inferSchema", "true")\
    .load("/Volumes/databrickscert/databrickscert/samplevolume/sales.csv")

display(df)
```

### read a csv without header, add schema 

```sql
%python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"
df_corrupted = spark.read.format("csv")\
    .schema(schema_str)\
    .option("inferSchema", "true")\
    .load("/Volumes/databrickscert/databrickscert/samplevolume/sales_no_header.csv.csv")

display(df_corrupted)
```

### read a csv without header, add schema 

```sql
%python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

_schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("customer_name", StringType(), True),
    StructField("product", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("price", DoubleType(), True),
    StructField("remarks", StringType(), True)
])

df2= spark.read.format("csv")\
    .schema(_schema)\
    .load("/Volumes/databrickscert/databrickscert/samplevolume/sales_no_header.csv")

display(df2)
```

### csv with single quote, escape, multiline

```sql
%python

df4= spark.read.format("csv")\
    .option("header", "true")\
    .option("sep", "|")\
    .option("quote", "'")\
    .option("escape", "#")\
    .option("multiline", "true")\
    .load("/Volumes/databrickscert/databrickscert/samplevolume/sales_pipe_single_quote_escape_multiline.csv")

display(df4)
       

```
